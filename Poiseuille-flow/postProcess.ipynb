```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_pressure= pd.read_csv('pressure.csv', header=None)
df_pressure.columns = ['X', 'Pressure']
df_pressure['Pressure'] = df_pressure['Pressure'] *1000
print(df_pressure.head(3))
print(df_pressure.shape)
```

```python

plt.style.use("dark_background")
plt.figure(figsize=(8,5))
df_pressure.plot(
    x='X', 
    y='Pressure', 
    kind='line', 
    color='yellow', 
    linewidth=2
)

plt.title('Pressure along X axis')
plt.xlabel('X')
plt.ylabel('Pressure')
plt.grid(True, which='both', axis='both', linestyle='--', linewidth=0.5, color='gray')
plt.show()
```

```python
u_df = pd.read_csv('velocity.csv',header=None)
u_df.columns = ['Height', 'Ux', 'Uy', 'Uz']
```

```python
import matplotlib.pyplot as plt

plt.style.use("dark_background")
plt.figure(figsize=(6,8))  # Taller figure for vertical profile

# Horizontal line plot: Velocity vs Height
plt.plot(u_df['Ux'], u_df['Height'], color='yellow', linewidth=2)

plt.title('Velocity along a vertical line at 0.75 m from inlet')
plt.xlabel('Velocity')
plt.ylabel('Height')
plt.grid(True, which='both', axis='both', linestyle='--', linewidth=0.5, color='gray')

plt.show()
```
